```python
import numpy as np
import matplotlib.pyplot as plt


degree = np.loadtxt("degree_list.txt")


plt.hist(degree, bins=20)

plt.xlabel("Degree k")
plt.ylabel("Number of nodes")

plt.title("Small-World Degree Distribution")

plt.savefig("Small-World _degree_distribution.png",
            dpi=300)

plt.show()
```
