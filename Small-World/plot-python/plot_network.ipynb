```python
import networkx as nx
import matplotlib.pyplot as plt


# Create graph

G = nx.Graph()


# Read edge list

with open("edge_list.txt","r") as file:

    for line in file:

        i,j = line.split()

        G.add_edge(int(i),int(j))


print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())



plt.figure(figsize=(8,8))


pos = nx.spring_layout(
    G,
    seed=1
)


nx.draw(
    G,
    pos,
    node_size=30,
    node_color="blue",
    edge_color="gray",
    alpha=0.5,
    with_labels=False
)


plt.title("Watts-Strogatz Small-World Network")


plt.savefig(
    "small_world_network.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()
```
