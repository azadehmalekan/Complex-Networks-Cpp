```python
import networkx as nx
import matplotlib.pyplot as plt


# Create empty graph

G = nx.Graph()


# Read edge list generated by C++

with open("scale_free_edge_list.txt", "r") as file:

    for line in file:

        i, j = line.split()

        G.add_edge(int(i), int(j))



# Draw network

plt.figure(figsize=(8,8))


pos = nx.spring_layout(G, seed=1)


nx.draw(
    G,
    pos,
    node_size=20,
    node_color="blue",
    edge_color="gray",
    with_labels=False
)


plt.title("Scale Free Random Network")


plt.savefig(
    "scale_free_network.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
